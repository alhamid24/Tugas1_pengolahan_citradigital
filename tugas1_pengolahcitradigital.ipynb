sebelum menganalisi gambar nya, disini kita harus menginstall pustaka yang kita butuhkan untuk menganalisis, disini saya akan menginstall pustaka Pillow dan NumPy pustaka ini berfungsi untuk membuktikan bahwa citra digital tersusun atas angka (piksel) sekaligus mengekstrak metadata dari file gambar tersebut.


```python
#import librery/pustaka
from PIL import Image
from PIL.ExifTags import TAGS
import numpy as np
```

#membuka citra menggunakan pillow
membuka citra menggunakan pillow atau pustaka yang telah kita import


```python
def analisis_citra(path_gambar):
    try:
        # 1. Membuka citra menggunakan Pillow
        img = Image.open(path_gambar)
        print("=" * 50)
        print(f"INFORMASI DASAR CITRA: {path_gambar}")
        print("=" * 50)
        print(f"- Format File : {img.format}")
        print(f"- Ukuran (WxH): {img.size[0]} x {img.size[1]} piksel")
        print(f"- Mode Warna  : {img.mode}")
        print("-" * 50 + "\n")

        # 2. Membuktikan bahwa citra terdiri dari angka/piksel (Matriks NumPy)
        # Mengubah objek gambar menjadi array multidimensi berisi nilai numerik (0-255)
        img_array = np.array(img)

        print("=" * 50)
        print("BUKTI: CITRA TERDIRI DARI ANGKA/PIKSEL (MATRIKS)")
        print("=" * 50)
        print(f"Dimensi Matriks (Shape): {img_array.shape}")
        print("Penjelasan dimensi: (Tinggi, Lebar, Kanal Warna)")
        print("\nContoh nilai angka pada piksel (sampel pojok kiri atas 3x3):")

        # Menampilkan sampel angka matriks piksel
        if len(img_array.shape) == 3:
            print(img_array[:3, :3, :]) # Untuk citra berwarna (RGB/RGBA)
        else:
            print(img_array[:3, :3])    # Untuk citra grayscale (keabuan)
        print("-" * 50 + "\n")

        # 3. Ekstraksi Metadata Citra
        print("=" * 50)
        print("EKSTRAKSI METADATA (EXIF)")
        print("=" * 50)

        exifdata = img.getexif()
        metadata_ditemukan = False

        if exifdata:
            for tag_id in exifdata:
                tag = TAGS.get(tag_id, tag_id)
                data = exifdata.get(tag_id)
                print(f"- {tag}: {data}")
                metadata_ditemukan = True

        if not metadata_ditemukan:
            print("Tidak ada metadata EXIF standar yang ditemukan pada gambar ini.")
            print("(Catatan: Gambar hasil tangkapan layar/screenshot atau editan seringkali tidak memiliki metadata EXIF).")

        print("=" * 50)

    except FileNotFoundError:
        print(f"Error: File '{path_gambar}' tidak ditemukan. Pastikan path file benar.")
    except Exception as e:
        print(f"Terjadi kesalahan: {e}")

# Jalankan fungsi dengan memasukkan nama atau path file gambar Anda
if __name__ == "__main__":
    # Ganti 'contoh_gambar.jpg' dengan nama file gambar yang ingin Anda uji
    file_target = "/content/drive/MyDrive/semester 5/pengolah citra digital/IMG_20260528_004651.jpg"
    analisis_citra(file_target)
```

```python
from PIL import Image
from PIL.ExifTags import TAGS, GPSTAGS
```

```python
def cek_lokasi_foto(path_gambar):
    image = Image.open(path_gambar)
    exif_data = image.getexif()

    if not exif_data:
        print("Tidak ada metadata EXIF pada foto ini.")
        return

    gps_info = None
    for tag_id, value in exif_data.items():
        tag = TAGS.get(tag_id, tag_id)
        if tag == "GPSInfo":
            gps_info = value

    if gps_info:
        print("[DITEMUKAN] Foto ini memiliki data GPS/Lokasi!")
        # Anda bisa memetakan tag GPS lebih lanjut di sini
    else:
        print("[TIDAK ADA] Foto ini tidak memiliki metadata lokasi GPS.")

# Contoh pemanggilan fungsi
# cek_lokasi_foto("nama_foto.jpg")
```
